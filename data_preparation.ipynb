```python
phantom_cols = [c for c in train.columns if c.startswith("Unnamed:")]
print(phantom_cols)
```

```python
train = train.drop(columns=phantom_cols)
```

```python
print(train.shape)
```

```python
def clean_col(c):
    return c.strip().replace(" ", "_")
```

```python
train.columns = [clean_col(c) for c in train.columns]
test.columns = [clean_col(c) for c in test.columns]
```

```python
print(list(train.columns) == list(test.columns))
```

```python
train["prognosis"].unique()
```

```python
train["prognosis"] = train["prognosis"].str.strip()
test["prognosis"] = test["prognosis"].str.strip()
```

```python
print(train["prognosis"].nunique())
```

```python
symptom_cols = [c for c in train.columns if c != "prognosis"]
```

```python
for c in symptom_cols:
    assert set(train[c].unique()) <= {0, 1}, f"{c} is not binary!"
```

```python
train[symptom_cols] = train[symptom_cols].astype("int8")
test[symptom_cols] = test[symptom_cols].astype("int8")
```

```python
train["prognosis"] = train["prognosis"].astype("category")
test["prognosis"] = test["prognosis"].astype("category")
```

```python
train.dtypes.value_counts()
```

```python
n_dupes = train.duplicated().sum()
print(n_dupes, "/", len(train))
```

```python
train.groupby("prognosis", observed=True)[symptom_cols].apply(lambda g: g.drop_duplicates().shape[0])
```

```python
train.to_csv("Training_clean.csv", index=False)
test.to_csv("Testing_clean.csv", index=False)

print("Saved.")
```
